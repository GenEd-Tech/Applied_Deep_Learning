#Word embedding

Word embedding is a deep learning technique that represents words as dense, real-valued vectors in a multi-dimensional space, where words with similar meanings have similar vector representations. This method allows deep learning models to understand the semantic and syntactic relationships between words, which is crucial for natural language processing (NLP) tasks like text classification and sentiment analysis

```python
import numpy as np

docs = [
    'go team go',
    'champions champions',
    'goal goal goal',
    'we will win this match',
    'defense defense defense',
    'messi messi',
    'ronaldo ronaldo',
    'neymar neymar',
    'let’s go boys',
    'victory is ours'
]
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(docs)
```

```python
len(tokenizer.word_index)
```

```python
tokenizer.word_index
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences

#go team go
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding

model = Sequential()
model.add(Embedding(17,output_dim=2))

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
model = Sequential()
model.add(Embedding(input_dim= 10000, output_dim=2))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
