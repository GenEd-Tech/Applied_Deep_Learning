```python
 model = models.Sequential()
 # Conv1
 model.add(layers.Conv2D(6, kernel_size=5, activation='tanh', input_shape=input_shape))  # -> 28x28x6
 model.add(layers.AveragePooling2D())                                                     # -> 14x14x6
 # Conv2
 model.add(layers.Conv2D(16, kernel_size=5, activation='tanh'))                          # -> 10x10x16
 model.add(layers.AveragePooling2D())                                                     # -> 5x5x16
 model.add(layers.Flatten())
 model.add(layers.Dense(120, activation='tanh'))
 model.add(layers.Dense(84, activation='tanh'))
 model.add(layers.Dense(num_classes, activation='softmax'))
```
