**Normalization** means rescaling values so they are on a similar scale.

In neural networks

Activations can become very large or very small

This causes:

* Slow training

* Unstable gradients

* Exploding / vanishing gradients

Normalization stabilizes training and speeds up convergence

Batch Normalisation

**Problems with Batch Normalisation**

Not suitable for sequence models

Transformers process variable-length sequences

Padding tokens distort batch statistics

# Layer Normalisation

Layer Normalization normalizes across the feature dimension, not the batch.

Hence No interference from:

Other sentences

Padding tokens
