#Optimizers

An optimizer is an algorithm used to update the weights of a neural network to minimize the loss function.


Or in even simpler words:

Optimizer = the brain that decides how to change weights so the model learns.

It uses the gradients (from backpropagation) + a strategy (rules like Momentum, AdaGrad, etc.) to move weights in the right direction.
